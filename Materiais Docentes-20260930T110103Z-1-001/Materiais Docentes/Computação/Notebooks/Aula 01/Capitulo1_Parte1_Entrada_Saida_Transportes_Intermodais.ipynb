## 2. A função input()
A função `input()` permite receber dados digitados pelo usuário.

```python
nome=input("Digite o nome do motorista: ")
print("Motorista cadastrado:",nome)
```

## 3. Variáveis
Uma variável armazena um valor para ser reutilizado.

```python
print("********* Lendo valores ********************************************")
empresa=input("Empresa: ")
origem=input("Origem: ")
destino=input("Destino: ")
print("********* Exibe valores separados por vírgula *********************")
print("Empresa:",empresa)
print("Origem:",origem)
print("Destino:",destino)
print("*******************************************************************")
print("********* Exibe valores concatenando ******************************")
print("Empresa: " + empresa)
print("Origem: " + origem)
print("Destino: " + destino)
print("*******************************************************************")
print("********* Exibe valores usando f-string ***************************")
print(f"Empresa: {empresa}")
print(f"Origem: {origem}")
print(f"Destino: {destino}")
print("*******************************************************************")
```

## Mini Projeto
Cadastre empresa, motorista, origem, destino e modal e exiba um resumo organizado.

```python
print("==========CADASTRO EMPRESARIAL==========")

empresa = input("Empresa: ")
motorista = input("Motorista: ")
origem = input("Origem: ")
destino = input("Destino: ")
modal = input("Modal: ")

print("Cadastro Concluído!✅")

print("===========RESUMO ORGANIZADO==========")

print(f"Transportadora: {empresa}")
print(f"Motorista: {motorista}")
print(f"Cidade de Origem: {origem}")
print(f"Cidade de Destino: {destino}")
print(f"Modal: {modal}")
```
