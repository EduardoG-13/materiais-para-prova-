### Exemplo 6 – Quais distâncias está acima da média dos valores da lista

Uma empresa registrou as distâncias percorridas em cinco viagens. Calcule e mostre a distância total, a média dos valores. No final, mostre os valores acima da média.


````python

# ==========================================================
# Exemplo 6
# Somando elementos da lista
# ==========================================================

distancias = [320, 410, 285, 150, 520]

total = 0

# Percorre todos os elementos da lista.
for distancia in distancias:

    # Acumula a distância atual.
    total += distancia

print(f"Distância total: {total} km")

# calcula e mostra a média
media = total / len(distancias)

print(f"Média das viagens: {media:.1f} km")
````

````python
# mostrar os elementos da lista que estão acima da média
for distancia in distancias:
    if distancia > media:
        print(f"{distancia} km acima da média")
````

Exemplo 7– Determinando o maior valor percorrendo a lista e sem usar funções da biblioteca Python.

Determine a maior distância percorrida.


````python

# ==========================================================
# Exemplo 7
# Maior distância
# ==========================================================

distancias = [320, 410, 285, 150, 520]

maior = distancias[0]

for distancia in distancias:

    if distancia > maior:

        maior = distancia

print(f"Maior distância: {maior} km")
````

### Exemplo 8 - Obter o maior elemento de uma lista usando função da biblioteca Python.

````python
# ==========================================================
# Exemplo
# Obtendo o maior elemento de uma lista utilizando max()
# ==========================================================

# Lista contendo as distâncias percorridas pelos caminhões (km).
distancias = [320, 410, 520, 285, 150]

print("=" * 50)
print("MAIOR DISTÂNCIA PERCORRIDA")
print("=" * 50)

# Exibe a lista de distâncias cadastradas.
print(f"Distâncias cadastradas: {distancias}")

# A função max() percorre todos os elementos da lista e
# retorna o maior valor encontrado.
maior_distancia = max(distancias)

# Exibe o resultado.
print(f"\nMaior distância: {maior_distancia} km")
````

Até agora utilizamos listas para armazenar conjuntos de informações. As listas possuem uma característica muito importante:
```
podem ser modificadas durante a execução do programa.
```
Entretanto, existem situações em que os dados não devem ser alterados.

Exemplos:
siglas dos estados brasileiros;

*   tipos de transporte;
*   meses do ano;
*   coordenadas geográficas;
*   categorias de contêineres;
*   tipos de modais.

Para representar esse tipo de informação utilizamos as **tuplas**.

Uma tupla é uma estrutura de dados que permite armazenar vários elementos em uma única variável.

Assim como uma lista, uma tupla:
*   armazena vários elementos;
*   possui índices;
*   mantém a ordem dos elementos;
*   permite elementos repetidos.

A principal diferença é:
```
Uma tupla não pode ser modificada depois de criada.
```

### O que é um conjunto?

Um conjunto (set) é uma estrutura de dados que armazena elementos únicos, ou seja, não permite elementos repetidos.

Além disso:
não possui índices;

*   não mantém uma ordem fixa dos elementos;
*   permite adicionar e remover elementos;
*   oferece operações matemáticas muito úteis.


Imagine as seguintes cidades visitadas pelos caminhões:
Santos
Campinas
Curitiba
Santos
Campinas
Vitória

Exemplo 13 - Ao armazenar essas informações em um conjunto teremos:


````python

cidades1 = {'Santos','Campinas','Curitiba','Vitória'}
cidades2 = {'São Paulo','Campinas','Santo André','Curitiba'}
print("União entre conjuntos - não há repetição")
print(cidades1 | cidades2)
print("Intersecção entre conjuntos")
print(cidades1 & cidades2)
````

Quando utilizar conjuntos?

Os conjuntos são recomendados quando desejamos:
*   eliminar elementos duplicados;
*   descobrir elementos em comum entre coleções;
*   identificar elementos exclusivos;
*   realizar pesquisas rápidas;
*   representar grupos de objetos.

Principais operações
|Operação	| Método |
|:----------------------------:|:----------------------------:|
|Criar conjunto	| set() |
|Adicionar elemento	| add() |
|Remover elemento	| remove() |
|Quantidade	| len() |
|Pesquisar |	in |
|União	| union() ou \| |
|Interseção	|intersection() ou &|
|Diferença	|difference() ou -|

### O que é um dicionário?

Um dicionário é uma estrutura composta por pares:
```
chave → valor
```
Cada informação é identificada por uma chave.

Exemplo 14:
Criamos um conteiner com dados sobre o código, origem, destino e peso.

````python
# define um novo conteiner
conteiner = {
"codigo":"CNT001",
"origem":"Santos",
"destino":"Roterdã",
"peso":28.5
}

# mostra valor a partir da chave
print("Código ", conteiner["codigo"])
print("Origem ", conteiner["origem"])
print("Destino ", conteiner["destino"])
print("Peso ", conteiner["peso"], " kg")
print("-------------------------------")
# adiciona novo par chave e valor
conteiner["transporte"]="Caminhão 1"
print("Conteiner completo ", conteiner)
print("-------------------------------")
# percorre o dicionário
for chave in conteiner:
    print(chave, " -> ", conteiner[chave])
print("-------------------------------")
# remove um par chave + valor
del conteiner["peso"]
print("Conteiner após o del ", conteiner)
print("-------------------------------")
# remove um par chave + valor
conteiner.pop("origem")
print("Conteiner após o pop ", conteiner)
print("-------------------------------")
print("Quantidade de pares ", len(conteiner))
print("-------------------------------")
print("Há a chave transporte em conteiner? ", 'transporte' in conteiner)

````
