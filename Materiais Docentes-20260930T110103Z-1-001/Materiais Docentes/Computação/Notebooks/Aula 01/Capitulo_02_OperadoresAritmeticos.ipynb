### Conceito - Precedência de Operadores

````python
# Cálculo de Peso de Carga Intermodal (Precedência de Operadores)
# Objetivo: Calcular o peso total de uma carga utilizando parênteses para precedência

# 1. Entrada de dados do usuário (aceita números inteiros ou decimais)
quantidade_caixas = float(input("Digite a quantidade de caixas a serem carregadas: "))
peso_produto = float(input("Digite o peso do produto (em kg) dentro de cada caixa: "))
peso_embalagem = float(input("Digite o peso da embalagem protetora (em kg) de cada caixa: "))
peso_palete = float(input("Digite o peso fixo do palete de madeira (em kg): "))

# 2. Cálculo do Peso Total utilizando parênteses para garantir a precedência correta.
# A soma (peso_produto + peso_embalagem) DEVE acontecer antes da multiplicação.
peso_total = quantidade_caixas * (peso_produto + peso_embalagem) + peso_palete # qual será o resultado se os parênteses forem retirados ?

# 3. Exibição do resultado formatado na tela
print(f"\nO peso total da carga é de {peso_total:.2f} kg.")
````

### Desafio 1
Faça a leitura do peso das cargas transportadas por três modais diferentes e calcule a carga total unificada em quilogramas.

Entrada - o peso da carga de:
1. dois caminhões (tonelada).
2. um trem (tonelada).
3. um avião (quilograma com pesagem direta da carga na balança).

Saída - Exiba o peso total transportado, convertido e somado em toneladas.

````python
caminhao_1 = float(input("Carga do 1º Caminhão em Toneladas: "))
caminhao_2 = float(input("Carga do 2º Caminhão em Toneladas: "))
trem = float(input("Carga do Trem em Toneladas: "))
aviao = float(input("Carga do Avião em Quilogramas: "))

conversao = aviao / 1000

soma = conversao + caminhao_1 + caminhao_2 + trem

print(f"O peso total transportado é de {soma} toneladas.")

capacidade_caminhao_1 = float(input("Capacidade total do 1º Caminhão em Toneladas: "))
capacidade_caminhao_2 = float(input("Capacidade total do 2º Caminhão em Toneladas: "))
capacidade_trem = float(input("Capacidade total do Trem em Toneladas: "))
capacidade_aviao = float(input("Capacidade total do Avião em Quilogramas: "))

conversao_capacidade = capacidade_aviao / 1000

restante_caminhao_1 = capacidade_caminhao_1 - caminhao_1
restante_caminhao_2 = capacidade_caminhao_2 - caminhao_2
restante_trem = capacidade_trem - trem
restante_aviao = conversao_capacidade - conversao

print(f"A Capacidade Restante do 1º Caminhão é de {restante_caminhao_1} Toneladas")
print(f"A Capacidade Restante do 2º Caminhão é de {restante_caminhao_2} Toneladas")
print(f"A Capacidade Restante do Trem é de {restante_trem} Toneladas")
print(f"A Capacidade Restante do Avião é de {restante_aviao} Toneladas")
````

### Desafio 2

Calcule quantos vagões são necessários para transportar uma grande safra de grãos (como soja ou milho) sem que nenhum vagão viaje incompleto.

Exemplo do Problema:
Volume total da safra: 4.250 toneladas de soja.
Capacidade de carga por Vagão do tipo Graneleiro: 70 toneladas.

O Cálculo com Divisão Inteira

4.250 // 70 = 60 com resto 50.

O operador ferroviário formará um trem com exatamente 60 vagões completamente cheios.
Sobra de Carga que Aguardará o Próximo Trem: Sobram 50 toneladas de soja no terminal de carga.

````python
carga_total = float(input("Volume total da Safra: "))

vagoes = carga_total // 70
resto = carga_total % 70

print("O operador ferroviário formará um trem com exatamente {} vagões completamente cheios.".format(vagoes))
print("Sobra de carga que aguardará o próximo trem: sobram {} toneladas de soja no terminal de carga.".format(resto))
````
