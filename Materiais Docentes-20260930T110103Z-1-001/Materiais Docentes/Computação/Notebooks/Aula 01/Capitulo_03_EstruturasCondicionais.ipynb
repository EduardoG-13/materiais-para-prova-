## Exemplo 2

**Objetivo:** Verificar se um caminhão pode iniciar a viagem (if...else)

Uma transportadora só autoriza a saída do caminhão se houver pelo menos 100 litros de diesel no tanque.

O programa deve ler a quantidade de combustível disponível e informar se a viagem pode ser iniciada.

### Casos de teste
|Entrada|Resultado|
|---|---:|
|99.9|Viagem não autorizada|
|100|Viagem autorizada|
|250|Viagem autorizada|
|0|Viagem não autorizada|

````python
# Quantidade mínima de combustível necessária para iniciar
# uma viagem.
COMBUSTIVEL_MINIMO = 100

# Lê a quantidade de combustível existente no tanque.
combustivel = float(input("Informe a quantidade de combustível (litros): "))

# Verifica se existe combustível suficiente.
if combustivel >= COMBUSTIVEL_MINIMO:
    # O caminhão possui combustível suficiente.
    print("\nViagem autorizada.")
    print(f"Combustível disponível: {combustivel:.1f} litros.")
else:
    # O caminhão não possui combustível suficiente.
    falta = COMBUSTIVEL_MINIMO - combustivel
    print("\nViagem NÃO autorizada.")
    print(f"Faltam {falta:.1f} litros para iniciar a viagem.")
````

### Erro comum

Esquecer de converter a entrada para número.

❌ **Incorreto**

```python
combustivel = input("Litros: ")
if combustivel >= 100:
    ...

Resultado
TypeError: '>=' not supported between instances of 'str' and 'int'
```

✅ **Correto**
```python
combustivel = float(input("Litros: "))
if combustivel >= 100:
    ...
```


## Exemplo 3

Objetivo: Verificar excesso de velocidade (if...else)

Em um terminal intermodal existe um limite máximo de velocidade de 20 km/h.

Desenvolva um programa que leia a velocidade de um caminhão e informe se ele está dentro do limite permitido.

Caso esteja acima do limite, informe também quanto excedeu.

````python
# Velocidade máxima permitida no terminal.
VELOCIDADE_MAXIMA = 20

# Lê a velocidade informada pelo usuário.
velocidade = float(input("Informe a velocidade do caminhão (km/h): "))

# Verifica se a velocidade está dentro do limite.
if velocidade <= VELOCIDADE_MAXIMA:
    # Caminhão dentro da velocidade permitida.
    print("\nVelocidade permitida.")
    print(f"Velocidade registrada: {velocidade:.1f} km/h")
else:
    # Calcula quanto a velocidade ultrapassou o limite.
    excesso = velocidade - VELOCIDADE_MAXIMA
    print("\nATENÇÃO!")
    print("Velocidade acima do permitido.")
    print(f"Excesso de {excesso:.1f} km/h.")
````

### Casos de teste
|Velocidade|Resultado|
|---|---:|
|15|Velocidade permitida|
|20|Velocidade permitida|
|20.1|Velocidade acima do permitido|
|35|Excesso de 15 km/h|
|0|Velocidade permitida|


### Erro comum

Utilizar < em vez de <=.

❌ Incorreto
```python
if velocidade < VELOCIDADE_MAXIMA:

Resultado: Nesse caso, uma velocidade exatamente igual a 20 km/h seria considerada incorretamente acima do limite.
```

✅ Correto
```python
if velocidade <= VELOCIDADE_MAXIMA:
```

### Exemplo 7 – Tabela Verdade do Operador OR

Uma carga poderá seguir viagem caso exista uma rota alternativa por:

*   ferrovia;
*   hidrovia.

Basta uma das alternativas estar disponível.

Implemente um programa que leia se há ou não uma rota alternativa para a ferrovia e para a hidrovia, e informe se a carga pode seguir viagem.

````python
# Solicita ao usuário se existe uma rota ferroviária disponível.
ferrovia = input("Existe rota ferroviária disponível? (S/N): ")

# Solicita ao usuário se existe uma rota hidroviária disponível.
hidrovia = input("Existe rota hidroviária disponível? (S/N): ")

# Remove espaços em branco e converte as respostas para letras maiúsculas.
ferrovia = ferrovia.strip().upper()
hidrovia = hidrovia.strip().upper()

# Verifica se as respostas informadas são válidas.
if (ferrovia not in ("S", "N")) or (hidrovia not in ("S", "N")):
    print("\nErro: informe apenas 'S' para Sim ou 'N' para Não.")
else:
    # O operador OR retorna verdadeiro quando
    # pelo menos uma das condições é verdadeira.
    if ferrovia == "S" or hidrovia == "S":
        print("\nA carga pode seguir viagem.")
        print("Existe pelo menos uma rota alternativa disponível.")
    else:
        print("\nA carga não pode seguir viagem.")
        print("Nenhuma rota alternativa está disponível.")
````

### Exemplo 8 – Tabela Verdade do Operador AND

Um caminhão somente poderá entrar no terminal logístico quando, ao mesmo tempo:

*   possuir autorização de acesso;
*   existir vaga disponível.

A entrada será permitida apenas quando as duas condições forem verdadeiras.

Implemente um programa que leia via teclado se o caminhão possui ou não autorização de acesso e se existe vaga disponível. O programa deve informar se o caminhão pode entrar ou não.

````python
# Solicita ao usuário se o caminhão possui autorização de acesso.
autorizacao = input("O caminhão possui autorização de acesso? (S/N): ")

# Solicita ao usuário se existe vaga disponível no terminal.
vaga = input("Existe vaga disponível? (S/N): ")

# Remove espaços em branco e converte a resposta para letras maiúsculas.
autorizacao = autorizacao.strip().upper()
vaga = vaga.strip().upper()

# Verifica se as respostas são válidas.
if (autorizacao not in ("S", "N")) or (vaga not in ("S", "N")):
    print("\nErro: informe apenas 'S' para Sim ou 'N' para Não.")
else:
    # O operador AND somente retorna True quando ambas
    # as condições são verdadeiras.
    if autorizacao == "S" and vaga == "S":
        print("\nEntrada autorizada.")
        print("O caminhão pode entrar no terminal.")
    else:
        print("\nEntrada não autorizada.")
        print("O caminhão não pode entrar no terminal.")
````

### Exemplo 9

O programa deve ler a resposta sobre se sua entrada está bloqueada ou não, e exibir uma resposta que informe ao usuário.

````python
# Solicita ao usuário se o caminhão está bloqueado.
bloqueado = input("O caminhão está bloqueado? (S/N): ")
# Remove espaços em branco (strip) e converte a resposta para letras maiúsculas (upper).
bloqueado = bloqueado.strip().upper()
# Verifica se a resposta é válida.
if bloqueado not in ("S", "N"):
    print("\nErro: informe apenas 'S' para Sim ou 'N' para Não.")
else:
    # Converte a resposta para um valor booleano.
    esta_bloqueado = bloqueado == "S"
    # O operador NOT inverte o valor lógico da variável.
    if not esta_bloqueado:
        print("\nEntrada permitida.")
        print("O caminhão não está bloqueado.")
    else:
        print("\nEntrada negada.")
        print("O caminhão está bloqueado.")
````
