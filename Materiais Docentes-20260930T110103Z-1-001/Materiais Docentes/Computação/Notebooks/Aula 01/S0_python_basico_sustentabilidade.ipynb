## 2. Tipos de dados

Em Python, todo valor tem um **tipo**. Os tipos mais básicos são:

| Tipo | Nome em Python | Exemplo |
|---|---|---|
| Número inteiro | `int` | `12` |
| Número decimal (ponto flutuante) | `float` | `12.5` |
| Texto | `str` | `"reciclagem"` |
| Verdadeiro/Falso | `bool` | `True`, `False` |

Podemos descobrir o tipo de um valor com a função `type()`.

````python
# definição dos valores para as variáveis
quantidade_sacolas = 30          # int
preco_combustivel = 5.89         # float
material = "plástico"            # str
casa_recicla = True               # bool

# exibe o tipo de cada variável
print(type(quantidade_sacolas))
print(type(preco_combustivel))
print(type(material))
print(type(casa_recicla))
````

### Conversão de tipos

Como vimos, o `input()` sempre devolve uma `str`. Se quisermos fazer contas, precisamos **converter** esse texto para número, usando `int()` (número inteiro) ou `float()` (número decimal).

````python
# ler a quilometragem rodada por semana
km_por_semana = input(nome + ", quantos km você roda de carro por semana? ")
print(type(km_por_semana))  # ainda é str!

# converter o valor lido para real
km_por_semana = float(km_por_semana)
print(type(km_por_semana))  # agora é float
````

Também podemos converter na hora, sem passar por uma variável intermediária:

````python
print(nome, ", por mês você rodará aproximadamente ", km_por_semana * 4.5, " km")  # já podemos fazer contas normalmente
````

## 3. Estrutura condicional (`if`, `elif`, `else`)

Muitas vezes um programa precisa tomar decisões diferentes dependendo dos dados. Para isso usamos a estrutura condicional `if` (se), `elif` (senão se) e `else` (senão).

Antes, vamos relembrar os principais **operadores de comparação**:

| Operador | Significado |
|---|---|
| `==` | igual a |
| `!=` | diferente de |
| `>` | maior que |
| `<` | menor que |
| `>=` | maior ou igual a |
| `<=` | menor ou igual a |

E os **operadores lógicos**: `and` (e), `or` (ou), `not` (negação).

<div class="alert alert-block alert-warning">
<b>Nota:</b>

Em Python, os blocos de código dentro de um `if` são definidos pela **indentação** (o espaço no início da linha), e não por chaves `{}` como em outras linguagens. Preste atenção nisso!
</div>

````python
consumo_kwh = float(input(nome + ", qual o consumo de energia da sua casa este mês (em kWh)? "))

if consumo_kwh < 150:
    print("Consumo baixo. Parabéns pela economia de energia!")
elif consumo_kwh < 300:
    print("Consumo médio. Dá para economizar um pouco mais.")
else:
    print("Consumo alto! Vale a pena revisar hábitos de uso de energia.")
````

Podemos combinar condições usando `and` / `or`. Por exemplo, um alerta ambiental pode depender de **dois** fatores ao mesmo tempo:

````python
print(nome, ", consulte a previsão do tempo e responda:")
temperatura = float(input("Temperatura de hoje (°C): "))
umidade = float(input("Umidade relativa do ar (%): "))

if temperatura > 35 and umidade < 20:
    print("Alerta de onda de calor e baixa umidade! Risco de incêndios.")
else:
    print("Sem alerta climático extremo por enquanto.")
````

## 4. Estrutura de repetição (`for` e `while`)

Loops (estruturas de repetição) servem para repetir um bloco de código várias vezes, sem precisar copiar e colar o mesmo código.

### `for`

O `for` é usado quando sabemos, de antemão, quantas vezes queremos repetir algo (ou sobre quais elementos queremos iterar). A função `range(n)` gera uma sequência de números de `0` até `n-1`.

````python
# Somando o consumo de água (em litros) ao longo de uma semana
consumo_total = 0

for dia in range(7):
    consumo_dia = float(input(f"Consumo de água no dia {dia + 1} (litros): "))
    consumo_total += consumo_dia

print("Consumo total na semana:", consumo_total, "litros")
print("Consumo médio diário:", consumo_total / 7, "litros")
````

### `while`

O `while` repete um bloco de código **enquanto** uma condição for verdadeira. É útil quando não sabemos, de antemão, quantas repetições serão necessárias.

````python
# Quantos anos até uma área reflorestada dobrar de tamanho, crescendo 8% ao ano?
area_atual = 10.0   # hectares
area_inicial = area_atual
anos = 0

while area_atual < 2 * area_inicial:
    area_atual = area_atual * 1.08
    anos += 1

print(f"Em {anos} anos a área reflorestada dobra de tamanho (~{area_atual:.2f} hectares).")
````

## 5. Listas

Uma lista é uma coleção **ordenada** de valores, que podem ser de qualquer tipo. Listas em Python são criadas com colchetes `[]`.

Você já viu listas no notebook de Numpy — agora vamos entender como elas funcionam em Python "puro" (sem a biblioteca Numpy).

````python
# Consumo de energia elétrica (kWh) nos últimos 6 meses
consumo_mensal = [320, 290, 310, 275, 260, 240]

print(consumo_mensal)
print("Número de meses registrados:", len(consumo_mensal))
````

### Acessando elementos

Assim como em arrays do Numpy, acessamos elementos de uma lista pelo índice (começando em `0`), e também podemos usar fatiamento (`slicing`).

````python
print("Consumo do primeiro mês:", consumo_mensal[0])
print("Consumo do último mês:", consumo_mensal[-1])
print("Consumo dos 3 primeiros meses:", consumo_mensal[:3])
````

### Funções e operações úteis com listas

* `len(lista)` — quantidade de elementos
* `sum(lista)` — soma dos elementos
* `max(lista)` / `min(lista)` — maior / menor valor
* `lista.append(valor)` — adiciona um elemento ao final
* percorrer com `for elemento in lista:`

````python
total = sum(consumo_mensal)
media = total / len(consumo_mensal)
maior = max(consumo_mensal)

print("Consumo total no período:", total, "kWh")
print("Consumo médio mensal:", media, "kWh")
print("Mês de maior consumo:", maior, "kWh")
````

````python
# Adicionando o consumo de um novo mês
consumo_mensal.append(255)
print(consumo_mensal)
````

````python
# Percorrendo a lista com for, mês a mês
for i, consumo in enumerate(consumo_mensal):
    print(f"Mês {i + 1}: {consumo} kWh")
````

## 6. Funções

Funções permitem **reaproveitar** um bloco de código, dando um nome a ele. Uma função pode receber **parâmetros** (entradas) e pode **retornar** um valor com a palavra-chave `return`.

A estrutura básica é:

```python
def nome_da_funcao(parametro1, parametro2):
    # bloco de código
    return resultado
```

````python
def calcular_pegada_carbono(km_carro, km_transporte_publico):
    # Calcula a pegada de carbono (em kg de CO2) de uma semana de deslocamentos.
    emissao_carro = km_carro * 0.12          # kg CO2 por km
    emissao_transporte = km_transporte_publico * 0.04  # kg CO2 por km
    return emissao_carro + emissao_transporte


pegada = calcular_pegada_carbono(km_carro=50, km_transporte_publico=20)
print(f"Pegada de carbono da semana: {pegada:.2f} kg de CO2")
````

Parâmetros também podem ter **valores padrão** (usados quando quem chama a função não informa aquele argumento):

````python
def classificar_consumo_agua(litros_por_dia, meta=150):
    if litros_por_dia <= meta:
        return "Dentro da meta"
    else:
        return "Acima da meta"


print(classificar_consumo_agua(120))
print(classificar_consumo_agua(200))
print(classificar_consumo_agua(200, meta=250))
````
